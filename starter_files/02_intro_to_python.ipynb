# Section 2: Modeling Information - The Basic Building Blocks

To analyze data, we first need a way to represent that data in our code. In this section, we'll start with the simplest forms of information – basic facts and values – and learn how Python understands them. This is the first step towards **Modeling Information**.

## 2.1 Literals & Core Data Types: Raw Information

The most basic way to represent information in Python is by using **literals**. Literals are fixed values written directly into your code. Think of these as the raw facts or pieces of data you might start with.

Each literal belongs to a fundamental **data type**. The data type tells Python what kind of information it is and what operations are allowed with it. Here are the core data types we'll use constantly:

* **Strings (`str`)**: Used for textual data. Enclose text in *either* single quotes (`'...'`) or double quotes (`"..."`).
    * Example: `"Midtown Office"`, `'Project Alpha'`
* **Integers (`int`)**: Used for whole numbers (positive, negative, or zero).
    * Example: `255`, `-100`, `2024`
* **Floats (`float`)**: Used for numbers that have a decimal component (floating-point numbers).
    * Example: `15750.25`, `0.05`, `3.14159` (Note: Even `2.0` is a float, not an int)
* **Booleans (`bool`)**: Represent logical `True` or `False` values. These are crucial for making decisions in your code. Note the capitalization.
    * Example: `True`, `False`
* **The `None` Type (`NoneType`)**: A special type representing the intentional absence of a value. It's often used as a placeholder or to indicate that data is missing or not applicable. There's only one `None` value.
    * Example: `None`

```python
# Examples of literals
print("This is a string.")
print('So is this, using single quotes.')
print("Company: FutureCorp")

print(42) # An integer
print(-150) # A negative integer

print(3.14159) # A float
print(2500.0)  # This is also a float because of the .0

print(True)  # A boolean
print(False) # Another boolean

print(None)  # The None value
```

**Checking the Data Type with `type()`**

Python provides a built-in function called `type()` that lets you ask what data type a specific value (or a variable holding a value) is. This is very useful for understanding your data and for debugging.

**Quick Tip for Debugging**: Remember `print()` and `type()`!

## 2.2 Expressions & Operators: Combining and Comparing Information

Rarely do we work with just single literal values. We usually need to combine, compare, or perform calculations on them. We do this using **expressions** and **operators**.

* An **expression** is a piece of code that Python evaluates to produce a value (e.g., `5 + 3` is an expression that evaluates to `8`).
* **Operators** are the special symbols that perform the actions (like `+`, `-`, `*`, `/`).

Here are common types of operators:

**Arithmetic Operators** (primarily for `int` and `float`):
* `+` : Addition
* `-` : Subtraction
* `*` : Multiplication
* `/` : True Division (result is *always* a float, e.g., `10 / 4` is `2.5`)
* `//`: Floor Division (discards the decimal/remainder, e.g., `10 // 4` is `2`)
* `%` : Modulo (gives the remainder, e.g., `10 % 4` is `2`)
* `**`: Exponentiation (e.g., `10 ** 2` is `100`)

**Comparison Operators** (these *always* result in a `True` or `False` boolean value):
* `==`: Equal to
* `!=`: Not equal to
* `<` : Less than
* `>` : Greater than
* `<=`: Less than or equal to
* `>=`: Greater than or equal to
    *(These also work for comparing strings alphabetically, e.g., `"apple" < "banana"` is `True`)*

**Logical Operators** (for combining `True`/`False` values):
* `and`: `True` if both sides are `True` (e.g., `True and True` is `True`)
* `or` : `True` if at least one side is `True` (e.g., `True or False` is `True`)
* `not`: Reverses the boolean value (e.g., `not True` is `False`)

```python
# Arithmetic Operator Examples
revenue = 50000
cost_of_goods = 30000
units_sold = 1000
marketing_spend = 5000

profit = revenue - cost_of_goods - marketing_spend
print(f"Profit: {profit}")

price_per_unit = revenue / units_sold
print(f"Price per unit: {price_per_unit}")

# What if we want to know how many full packs of 12 we can make from 150 items?
items = 150
pack_size = 12
full_packs = items // pack_size
remaining_items = items % pack_size
print(f"Full packs of {pack_size}: {full_packs}, Remaining items: {remaining_items}")

# Growth over 2 years at 10% annually (1.1^2)
initial_investment = 1000
growth_factor = 1.1 ** 2
future_value = initial_investment * growth_factor
print(f"Future value: {future_value}")
```

```python
# Comparison Operator Examples
min_profitable_sales = 40000
actual_sales_q1 = 45000
actual_sales_q2 = 38000

is_q1_profitable = actual_sales_q1 > min_profitable_sales
print(f"Were Q1 sales profitable? {is_q1_profitable}")
print(f"Type of is_q1_profitable: {type(is_q1_profitable)}") # It's a boolean!

is_q2_meeting_target = actual_sales_q2 >= min_profitable_sales
print(f"Did Q2 sales meet the minimum target? {is_q2_meeting_target}")

are_sales_consistent = actual_sales_q1 == actual_sales_q2
print(f"Were Q1 and Q2 sales identical? {are_sales_consistent}")
```

```python
# Logical Operator Examples
has_high_revenue = actual_sales_q1 > 42000  # Let's say this is True (45000 > 42000)
has_low_costs = cost_of_goods < 32000    # Let's say this is True (30000 < 32000)
is_new_product = False

# Eligible for bonus if high revenue AND low costs
eligible_for_bonus = has_high_revenue and has_low_costs
print(f"Eligible for performance bonus? {eligible_for_bonus}")

# Flag for review if it's a new product OR if not meeting sales targets
flag_for_review = is_new_product or (not is_q2_meeting_target) # (not False is True)
print(f"Flag product for review? {flag_for_review}")

# Using 'not'
print(f"Is Q2 NOT meeting target? {not is_q2_meeting_target}")
```

## 2.3 Variables: Naming Information for Reuse

Hardcoding literal values everywhere in your analysis makes your code hard to read, understand, and maintain. If a value (like a tax rate or a sales target) needs to change, you'd have to find and replace it in many places!

More importantly for our "Modeling Information" approach, we want to give meaningful names to the different pieces of our information model. We achieve this using **variables**.

* **Assignment (`=`)**: You create a variable and assign it a value using the single equals sign (`=`).
    `variable_name = value`
* **Analogy**: Think of a variable as a label or a tag pointing to a value stored in the computer's memory. You use the label (the variable name) to refer to that value.

Once you assign a value to a variable, you can use the variable name in expressions, print it, or use it anywhere you would use the value itself.

```python
# Modeling details for a product
product_id = "PROD007"         # Assigning a string
product_name = "Wireless Mouse X2" # Assigning another string
units_in_stock = 150            # Assigning an int
unit_price = 29.99              # Assigning a float
is_active_product = True        # Assigning a bool

# Using the variables
print(product_name)
print(f"Price per unit: {unit_price}")

# Variables can be used in expressions
total_stock_value = units_in_stock * unit_price
print(f"Total stock value for {product_name} (ID: {product_id}): ${total_stock_value}")

# Variable values can be updated (re-assignment)
print(f"Units in stock before sale: {units_in_stock}")
units_sold_today = 25
units_in_stock = units_in_stock - units_sold_today # Update the variable
print(f"Units in stock after selling {units_sold_today}: {units_in_stock}")
```

**Variable Naming Conventions (Python Best Practice - PEP 8)**:

* **Use `snake_case`**: All lowercase letters, with words separated by underscores (e.g., `hourly_rate`, `customer_first_name`, `net_profit_margin`). This is the standard convention in the Python community and makes code more readable.
* **Be Descriptive**: Choose names that clearly indicate the variable's purpose (e.g., `average_customer_rating` is much better than `acr` or `rating_val`).
* **Start with a letter or underscore**: Variable names must start with a letter (a-z, A-Z) or an underscore (`_`). They can then be followed by letters, numbers (0-9), or underscores.
* **Case-Sensitive**: `my_variable` is different from `My_Variable`.
* **Avoid Python Keywords**: Do not use Python's reserved keywords (like `if`, `for`, `while`, `True`, `None`, `def`, `class`, etc.) as variable names.

**Comments (`#`): Explaining Your Code**

As you assign variables and write logic, it's crucial to add **comments** to your code. In Python, anything on a line after the hash symbol (`#`) is a comment and is ignored by the Python interpreter.

Comments don't affect how your code runs, but they are vital for:
* Explaining the *purpose* or *why* behind a piece of code.
* Making your code understandable to others (and to your future self!).
* Temporarily "turning off" lines of code during debugging (known as "commenting out").

## 2.4 Working with Text (`str`): A Closer Look

Text data (strings, or `str` in Python) is fundamental in business. You'll encounter it everywhere: customer names, product descriptions, addresses, survey responses, log files, financial report narratives, and much more. Python's strings come with powerful built-in capabilities for manipulation, which is often a key part of data cleaning and preparation.

**Combining Strings (Concatenation)**:
You can combine strings using the `+` operator. Remember to add spaces manually if needed!

**Common String Methods**:
Strings are objects in Python, and objects have built-in functions called **methods** that perform actions related to the object. You call a method using *dot notation*: `your_string_variable.method_name(arguments)`.

Here are some essential string methods for data cleaning and manipulation:
* `.lower()`: Returns a *new* string with all characters converted to lowercase.
* `.upper()`: Returns a *new* string with all characters converted to uppercase.
* `.strip()`: Returns a *new* string with leading and trailing whitespace (spaces, tabs, newlines) removed. This is invaluable for cleaning user input or data from files!
* `.replace(old, new)`: Returns a *new* string where all occurrences of the `old` substring are replaced with the `new` substring.
* `.startswith(prefix)`: Returns `True` if the string starts with the `prefix`, `False` otherwise.
* `.endswith(suffix)`: Returns `True` if the string ends with the `suffix`, `False` otherwise.
* `.split(separator)`: Returns a *list* of substrings that were separated by the `separator`. If no separator is given, it splits by whitespace.

**Important Note on Immutability**: Strings in Python are **immutable**. This means that string methods *do not change the original string*. Instead, they return a *new* string with the modifications. If you want to keep the change, you must assign the result back to a variable (it can be the same variable or a new one).

```python
# String Concatenation
first_name = "Jane"
last_name = "Doe"
role = "Analyst"

full_name = first_name + " " + last_name # Don't forget the space!
print(f"Full Name: {full_name}")

# String Methods
messy_department_code = "  FINANCE_DEPT_US\n" # Has extra spaces and a newline
print(f"Original Department Code: '{messy_department_code}'")

# Clean it up
clean_department_code = messy_department_code.strip().upper()
print(f"Cleaned Department Code: '{clean_department_code}'")

# Using replace
report_title = "Quarterly Report - Q2 2024"
updated_report_title = report_title.replace("2024", "2025")
print(f"Original Title: {report_title}")
print(f"Updated Title: {updated_report_title}") # Original report_title is unchanged

# Using startswith and endswith
file_name = "SalesData_Final_Report.xlsx"
if file_name.startswith("SalesData") and file_name.endswith(".xlsx"):
    print(f"'{file_name}' looks like a valid sales report Excel file.")
else:
    print(f"'{file_name}' might not be the correct file type or naming convention.")

# Using split
tags_string = "marketing, finance, strategy, operations"
tags_list = tags_string.split(", ") # Split by comma followed by a space
print(f"Tags as a list: {tags_list}")
print(f"The first tag is: {tags_list[0]}") # We'll learn about lists soon!
```

**F-Strings (Formatted String Literals): The Best Way to Combine Strings and Variables**

We've already used f-strings in many of our `print()` examples! They provide the clearest, most readable, and most convenient way to create strings that include the values of variables.

**How to use f-strings**:
1.  Start the string with the letter `f` or `F` immediately before the opening quote (`f"..."` or `f'...'`).
2.  Inside the string, place variable names or even simple Python expressions directly within curly braces `{}`. Python will replace the braces and their contents with the current value of the variable or the result of the expression.

You can also control the formatting of numbers within f-strings. For example, `:.2f` is often used to format a float to display exactly two decimal places (very useful for currency).

```python
# F-String Examples
product_id = "SRV003"
product_name = "Cloud Storage Subscription"
monthly_cost = 49.99
users = 150
discount_percentage = 10 # As a whole number

# Basic f-string usage
output_message = f"Product: {product_name} (ID: {product_id}) costs ${monthly_cost} per month."
print(output_message)

# F-strings with calculations
total_annual_cost = monthly_cost * 12
print(f"The total annual cost for {product_name} is ${total_annual_cost:.2f}.") # Formatted to 2 decimal places

discounted_monthly_cost = monthly_cost * (1 - discount_percentage / 100)
print(f"With a {discount_percentage}% discount, the monthly cost is ${discounted_monthly_cost:.2f}.")

# You can even include more complex expressions or function calls (though keep it readable)
active_status_text = "Active" if is_active_product else "Inactive" # This is a conditional expression
print(f"The product '{product_name}' is currently {active_status_text.upper()}.") # Method call inside f-string
```

## 3.1 Importing Modules: Accessing More Tools

Python itself provides the basic building blocks, but much of its power, especially for data analysis, comes from **modules** and **packages** (often collectively called **libraries**). These are collections of pre-written code, created by other developers (or by Python's core team), that provide specialized functions, data types, and capabilities.

Think of Python's built-in features as your standard workshop tools. Modules are like specialized toolsets you can bring into your workshop when needed:
* A set for advanced mathematical operations.
* A set for creating charts and visualizations.
* A set for accessing web data.
* Importantly for us, tools for handling dates and times, or performing large-scale data manipulation (like the Polars library we'll see later).

To use the tools (functions, classes, etc.) from a module or library, you must first **`import`** it into your current Colab session (or Python script).

**How to Import:**

1.  **`import module_name`**: This brings the entire module into your session's memory. To use something *from* the module, you typically prefix it with the module's name followed by a dot (`.`).
    * Example: `math.sqrt(25)` would use the `sqrt` function from the `math` module.

2.  **`import module_name as alias`**: This is extremely common, especially for modules with long names or those that have widely accepted standard abbreviations. It imports the module but assigns it a shorter, convenient nickname (**alias**) that you use instead of the full name. This makes your code less verbose and often follows community conventions.
    * Example: `import datetime as dt` (common for the `datetime` module).
    * Example: `import polars as pl` (which we will use extensively later).

**Convention for Imports**:
By strong convention, all `import` statements are usually placed at the **very top** of a Jupyter Notebook or Python script. This clearly declares all the external tools your code relies on from the outset.

```python
# Example of importing and using a module
# The 'datetime' module provides tools for working with dates and times.
# We import it using a common alias 'dt'.

import datetime as dt

# Get today's date
todays_date = dt.date.today()
print(f"Today's date is: {todays_date}")

# Date objects have attributes you can access
print(f"The current year is: {todays_date.year}")
print(f"The current month is: {todays_date.month}")
print(f"The current day is: {todays_date.day}")

# We can also create specific date objects
specific_date = dt.date(2025, 1, 31) # Year, Month, Day
print(f"A specific date: {specific_date}")
print(f"The weekday of {specific_date} is: {specific_date.weekday()} (Monday=0, Sunday=6)")

# Get the current date and time
current_timestamp = dt.datetime.now()
print(f"Current date and time: {current_timestamp}")
print(f"Just the hour: {current_timestamp.hour}")
print(f"Just the minute: {current_timestamp.minute}")
```

## 3.2 Lists (`list`): Ordered, Changeable Sequences

**Modeling Motivation**: How do you represent a collection of related items where the *order is important* and the items might need to change?
* A sequence of steps in a marketing campaign.
* Monthly revenue figures recorded in chronological order.
* A list of client names, perhaps in the order they were acquired.
* The columns you want to select from a dataset.

Python's solution is the **list**. A list is an **ordered sequence** of items.
* Items can be of any data type (though often they are homogeneous, meaning all the same type, especially in data analysis).
* The order in which you add items is preserved.
* Lists are **mutable**, meaning you can change their contents after they are created (add, remove, or modify items).

**Syntax**: Create lists using square brackets `[]`, with items separated by commas.

```python
# A list of region names (strings)
regions = ["North", "South", "East", "West", "Central"]
print(f"Regions: {regions}")

# A list of quarterly sales targets (floats)
quarterly_sales_targets = [150000.00, 182000.50, 141000.75, 195000.20]
print(f"Quarterly Sales Targets: {quarterly_sales_targets}")

# A list can technically mix data types (but use with caution for data analysis)
# For data analysis, lists (which often become columns) usually contain the same type of data.
mixed_info = ["Product Launch Plan", 2025, 15000.00, True]
print(f"Mixed Info List: {mixed_info}")

# An empty list, ready to be filled later
pending_tasks = []
print(f"Pending Tasks (initially): {pending_tasks}")
```

**Accessing Items by Index (Position)**:
Because lists are ordered, you retrieve items using their numerical position, called an **index**.
* **Crucially, Python indexing starts from 0!** The first item is at index `0`, the second at index `1`, and so on.
* You can also use **negative indexing**: `-1` refers to the last item, `-2` to the second-to-last, etc.

Trying to access an index outside the list's range (e.g., index `10` in a list of 5 items) will cause an `IndexError`.

```python
# Accessing items from our 'regions' list: ["North", "South", "East", "West", "Central"]
first_region = regions[0]         # Index 0 is the FIRST item
second_region = regions[1]        # Index 1 is the SECOND item

print(f"First region: {first_region}")
print(f"Second region: {second_region}")

last_region = regions[-1]         # Index -1 is the LAST item
second_to_last_region = regions[-2] # Index -2 is the second-to-last

print(f"Last region: {last_region}")
print(f"Second-to-last region: {second_to_last_region}")

# Let's try to access an out-of-range index (this will produce an error)
# print(regions[10]) # Uncomment this line to see the IndexError
```

**Slicing (Getting Sub-Lists)**:
You can extract a portion (a "slice" or sub-list) of a list using the syntax `my_list[start:stop]`.
* The `start` index is the position of the first item you want to include.
* The `stop` index is the position of the first item you *do not* want to include (i.e., slicing goes up to, but does not include, the `stop` index).
* If `start` is omitted, slicing starts from the beginning of the list (`my_list[:stop]`).
* If `stop` is omitted, slicing goes to the end of the list (`my_list[start:]`).
* You can also use a `step`: `my_list[start:stop:step]`.

```python
# Slicing our 'regions' list: ["North", "South", "East", "West", "Central"]

# Get 'South' and 'East' (index 1 up to, but not including, index 3)
middle_regions = regions[1:3]
print(f"Middle regions (index 1 to 3 exclusive): {middle_regions}")

# Get from the beginning up to index 2 (exclusive)
first_two_regions = regions[:2]
print(f"First two regions (up to index 2 exclusive): {first_two_regions}")

# Get from index 2 to the end
regions_from_east_onwards = regions[2:]
print(f"Regions from 'East' onwards (index 2 to end): {regions_from_east_onwards}")

# Get every other region, starting from the first
every_other_region = regions[::2] # Start defaults to 0, Stop defaults to end
print(f"Every other region: {every_other_region}")

# Get a copy of the entire list
all_regions_copy = regions[:]
print(f"A copy of all regions: {all_regions_copy}")
```

**Mutability (Lists are Changeable)**:
Lists are **mutable**, meaning you can change their contents after they are created.
* **Change an item**: Use index assignment (`my_list[index] = new_value`).
* **Add an item to the end**: Use the `.append()` method.
* There are other methods like `.insert()` to add at a specific position, `.remove()` to delete an item by value, and `del my_list[index]` to delete by index.

**Getting the Length of a List**:
Use the built-in `len()` function to find out how many items are in a list.

## 3.3 Tuples (`tuple`): Ordered, Unchangeable Records

**Modeling Motivation**: What if you want to group related pieces of information about a *single thing*, where the order matters, but the structure itself *should not change* after it's created?
* An (X, Y) coordinate point on a map or chart.
* An RGB color value: `(red_value, green_value, blue_value)`.
* A basic, fixed record for an employee: `(employee_id, name, hire_date)`.
* A specific version number `(major, minor, patch)`.

Python provides **tuples** for this. A tuple is also an **ordered sequence**, much like a list.

**Key Difference**: Tuples are **immutable**. Once created, you *cannot* change, add, or remove elements. This immutability signals that the tuple represents a fixed record or structure.

**Syntax**: Usually created using parentheses `()`, with items separated by commas.
(Parentheses are often optional if the context is clear, but it's good practice to use them for readability).
* A tuple with a single item *must* have a trailing comma: `(item,)`.

```python
# Tuples examples

# Tuple representing (latitude, longitude) coordinates for a store
store_location_A = (40.7128, -74.0060) # (Latitude, Longitude) for NYC
store_location_B = (34.0522, -118.2437) # for LA
print(f"Store A Coordinates: {store_location_A}")

# Tuple representing a fixed financial period (Year, Quarter)
fiscal_period = (2024, "Q3")
print(f"Fiscal Period: {fiscal_period}")

# Tuple representing a product version
software_version = (2, 1, 5) # Major, Minor, Patch
print(f"Software Version: {software_version}")

# A tuple requires a comma even for a single item!
# This distinguishes it from just a value in parentheses (which would be for operator precedence).
single_config_setting = ("localhost",)
print(f"Single config setting (tuple): {single_config_setting}")
print(f"Type of single_config_setting: {type(single_config_setting)}")

# Compare to this:
not_a_tuple = ("localhost") # This is just a string
print(f"Not a tuple (string): {not_a_tuple}")
print(f"Type of not_a_tuple: {type(not_a_tuple)}")


# Empty tuple
empty_record = ()
print(f"Empty record: {empty_record}")
```

**Accessing Items by Index (Same as Lists)**:
You access items in a tuple using their zero-based index, just like lists.

**Immutability (The Defining Feature)**:
This is where tuples differ critically from lists. Once a tuple is created, its contents cannot be altered.

```python
# Accessing items in a tuple
print(f"Store A Latitude: {store_location_A[0]}")
print(f"Store A Longitude: {store_location_A[1]}")

print(f"Software Major Version: {software_version[0]}")

# --- Attempting to change a tuple will cause an ERROR ---
# If you uncomment and run the following lines, you'll get a TypeError.

# store_location_A[0] = 40.7129  # This will raise a TypeError: 'tuple' object does not support item assignment
# software_version.append(3)     # This will raise an AttributeError: 'tuple' object has no attribute 'append'

# print("This line will not be reached if the errors above are uncommented.")
```

**Why Use Tuples?**

1.  **Data Integrity**: They signal that this group of items represents a fixed record or structure that shouldn't be accidentally altered. If you have a collection of items that should logically stay together and not change, a tuple is a good choice.
2.  **Performance (Minor)**: Tuples can be slightly more memory-efficient and sometimes faster to process than lists for certain operations, though this is usually not a primary concern for basic usage.
3.  **Dictionary Keys**: Because they are immutable, tuples can be used as keys in dictionaries (which we'll see next). Lists, being mutable, cannot be dictionary keys.

**Common Use Case: List of Tuples**
A very frequent and useful pattern for representing simple tables of data (like rows from a spreadsheet or database) is a **list of tuples**. Each tuple in the list represents one row or record, and its elements represent the fields of that record.

```python
# List of tuples: (Product ID, Product Name, Unit Price, Quantity in Stock)
inventory = [
    ("P1001", "Laptop Pro 15", 1299.99, 50),
    ("P1002", "Wireless Mouse XG", 24.50, 120),
    ("P1003", "External SSD 1TB", 175.00, 75),
    ("A2005", "Office Software Suite", 199.00, 200) # Another product
]
print("Current Inventory (List of Tuples):")
print(inventory)

# Access the second product record (which is a tuple)
product2_record = inventory[1] # inventory[1] is the tuple ("P1002", "Wireless Mouse XG", ...)
print(f"\nProduct 2 Record: {product2_record}")

# Access the unit price of the second product
# First, get the tuple for the second product (inventory[1])
# Then, get the price from that tuple (index 2)
product2_price = inventory[1][2]
print(f"Product 2 Unit Price: ${product2_price}")

# Access the name of the third product
product3_name = inventory[2][1]
print(f"Product 3 Name: {product3_name}")

# We can easily add another record (tuple) to our list of inventory
new_product_record = ("C3010", "Webcam HD Pro", 65.00, 90)
inventory.append(new_product_record) # Appending to the LIST, not the tuples themselves
print("\nInventory after adding a new product:")
print(inventory)
```

## 3.4 Dictionaries (`dict`): Labeled Information (Key-Value Pairs)

**Modeling Motivation**: Accessing items by numerical position (index) like in lists and tuples isn't always the most intuitive or practical way to retrieve information.
* How would you model configuration settings where you want to look up the 'server_ip_address' or 'default_timeout_seconds'?
* How would you represent a product where you want to directly access its 'price' or 'brand_name' or 'sku' using those specific names?

Python's **dictionaries** (or `dict`) are perfect for this. A dictionary stores a collection of **key-value pairs**.
* Each unique **key** is associated with a specific **value**.
* Think of it like a real-world dictionary where the word (key) points to its definition (value), or like a labeled filing cabinet.
* Keys must be unique within a dictionary.
* Keys must be of an immutable type (e.g., strings, numbers, or tuples are common keys; lists cannot be keys).
* Values can be of any data type (including other lists or dictionaries!).
* Dictionaries are **mutable**: you can add, remove, or change key-value pairs after creation.
* In modern Python (3.7+), dictionaries also remember the order in which items were inserted.

**Syntax**: Dictionaries are created using curly braces `{}`, containing `key: value` pairs separated by commas.

```python
# Dictionary modeling application configuration settings
app_config = {
    "server_ip": "192.168.1.100",
    "port": 8080,
    "api_version": "v2.1",
    "default_timeout_seconds": 60,
    "active_feature_flags": ["new_dashboard", "beta_analytics_panel"], # Value can be a list!
    "retry_attempts": 3
}
print(f"Application Configuration: {app_config}")

# Dictionary modeling a customer record
customer_profile = {
    "customer_id": "CUST-00789",
    "name": "Alice Wonderland",
    "email": "alice.w@example.com",
    "loyalty_tier": "Gold",
    "total_spend_usd": 1250.75,
    "is_subscribed_to_newsletter": True
}
print(f"\nCustomer Profile: {customer_profile}")

# Empty dictionary, can be filled later
project_details = {}
print(f"\nEmpty Project Details: {project_details}")
```

**Accessing Values by Key**:
You don't use numerical indices with dictionaries. Instead, you use the **key** inside square brackets `[]` to get its associated value.

Attempting to access a key that doesn't exist in the dictionary will raise a `KeyError`.
A safer way to access keys that might be missing is using the `.get()` method: `my_dict.get('some_key', default_value_if_missing)`.

```python
# Accessing values from our dictionaries

server_address = app_config["server_ip"]
print(f"Server IP from config: {server_address}")

customer_name = customer_profile["name"]
customer_email = customer_profile["email"]
print(f"Customer: {customer_name}, Email: {customer_email}")

# What happens if we try to access a key that doesn't exist?
# print(app_config["admin_username"]) # Uncomment to see a KeyError

# Safer access using .get()
# .get() returns None if the key is not found, or you can provide a default value.
admin_username = app_config.get("admin_username")
print(f"Admin Username (using .get()): {admin_username}") # Will be None

admin_username_with_default = app_config.get("admin_username", "default_admin")
print(f"Admin Username (with default): {admin_username_with_default}")

# Accessing the list of feature flags
active_flags = app_config["active_feature_flags"]
print(f"Active feature flags: {active_flags}")
print(f"The first active feature flag is: {active_flags[0]}") # This is a list, so we use list indexing
```

**Mutability (Dictionaries are Changeable)**:
Dictionaries are **mutable**. You can:
* Add new key-value pairs.
* Change the value associated with an existing key.
* Remove key-value pairs (e.g., using `del my_dict['key_to_delete']` or the `.pop()` method).

```python
# Modifying dictionaries
print(f"Original Customer Profile: {customer_profile}")

# Update the customer's email and loyalty tier
customer_profile["email"] = "alice.wonderland@newdomain.com" # Updates existing key
customer_profile["loyalty_tier"] = "Platinum"             # Updates existing key
print(f"\nUpdated Customer Profile (email, tier): {customer_profile}")

# Add new information: a phone number
customer_profile["phone_number"] = "555-123-4567" # Adds a new key-value pair
print(f"\nCustomer Profile after adding phone: {customer_profile}")

# Add details to the 'project_details' dictionary
project_details["project_name"] = "Q4 Market Expansion Analysis"
project_details["due_date"] = dt.date(2025, 10, 15) # Value can be a date object!
project_details["status"] = "In Progress"
print(f"\nUpdated Project Details: {project_details}")

# Removing a key-value pair
# Let's say we no longer want to store 'retry_attempts' in app_config
if "retry_attempts" in app_config: # Good practice to check if key exists before deleting
    del app_config["retry_attempts"]
print(f"\nApp Config after deleting 'retry_attempts': {app_config}")
```

**Key vs. Index: The Fundamental Difference in Retrieval**

This is a crucial distinction:
* **Lists/Tuples**: Are ordered sequences. You access items by their numerical **position (index)**.
    * Example: `my_list[0]`, `my_tuple[2]`
* **Dictionaries**: Store data as key-value pairs. You access values by their unique **label (key)**. They are conceptually unordered for access purposes (though modern Python preserves insertion order for iteration).
    * Example: `my_dict['setting_name']`, `my_dict['customer_id']`

Dictionaries are incredibly flexible and widely used for modeling objects or records where accessing information by a specific name or identifier is more natural and efficient than relying on numerical position.
